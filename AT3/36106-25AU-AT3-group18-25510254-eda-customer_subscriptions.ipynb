```python
#we aggregated rows which has the same customer_id
df_cleaned = pd.concat([agg_num, agg_cat], axis=1).reset_index()
```

```python
#aggregate duplicated rows
numerical_cols = df_details.select_dtypes(include=['int64', 'float64']).columns.tolist()
categorical_cols = df_details.select_dtypes(include=['object']).columns.tolist()
categorical_cols.remove('customer_id')

agg_num = df_details.groupby('customer_id')[numerical_cols].mean()
agg_cat = df_details.groupby('customer_id')[categorical_cols].agg(lambda x: x.mode().iloc[0] if not x.mode().empty else x.iloc[0])

df_details_cleaned = pd.concat([agg_num, agg_cat], axis=1).reset_index()
```

### C.2 Fixing "handling missing values"

> Provide some explanations on why you believe it is important to fix this issue and its impacts

> You can add more cells in this section

```python
# <Student to fill this section>
# List columns with missing values, count, and percent
missing_info = df_clean.isnull().sum()
missing_info = missing_info[missing_info > 0].sort_values(ascending=False)

missing_df = missing_info.to_frame().reset_index()
missing_df.columns = ['feature', 'missing_count']
missing_df['missing_percent'] = (missing_df['missing_count'] / len(df)) * 100

print(missing_df)
```

```python
# Binary/Yes-No to 0/1
binary_map = {'Yes': 1, 'No': 0}
df_clean['unlimited_data'] = df_clean['unlimited_data'].map(binary_map).fillna(0)

# Fill categorical with 'Unknown' or mode
df_clean['plan'] = df_clean['plan'].fillna('Unknown')
df_clean['broadband_type'] = df_clean['broadband_type'].fillna('Unknown')
df_clean['payment_method'] = df_clean['payment_method'].fillna(df_clean['payment_method'].mode()[0])

# Fill numeric with median or 0
df_clean['total_revenue'].fillna(df_clean['total_revenue'].median(), inplace=True)
df_clean['total_charges'].fillna(df_clean['total_charges'].median(), inplace=True)
df_clean['month_tenure'].fillna(df_clean['month_tenure'].median(), inplace=True)
df_clean['total_international_charges'].fillna(0, inplace=True)
df_clean['total_extra_data_charges'].fillna(0, inplace=True)
df_clean['total_refunds'].fillna(0, inplace=True)
```

```python
missing_info = df_clean.isnull().sum()
missing_info = missing_info[missing_info > 0].sort_values(ascending=False)

missing_df = missing_info.to_frame().reset_index()
missing_df.columns = ['feature', 'missing_count']
missing_df['missing_percent'] = (missing_df['missing_count'] / len(df)) * 100

print(missing_df)

#all of the features with missing values already filled.
```

```python
# <Student to fill this section>
data_cleaning_1_explanations = """
Clustering algorithms cannot handle missing values natively and rely heavily on distance-based calculations.
Unfilled missing values can distort distances or lead to dropped rows, which may bias the results or reduce data
coverage.
the approaches were selected based on types of features, numerical values filled with median or 0. categorical values
filled with unknown, and binay values filled with 0.
"""
```
