### C.4 Fixing "Convert and clean 'postcode'"

> You can add more cells in this section

```python
# <Student to fill this section>
before_fix = df_clean['postcode'].isnull().sum()
df_clean['postcode'] = df_clean['postcode'].fillna(-1).astype(int)
after_fix = df_clean['postcode'].isnull().sum()

print(f"Missing values in 'postcode' before: {before_fix}, after: {after_fix}")
```

```python
# <Student to fill this section>
data_cleaning_3_explanations = """
Although no missing values were present in the 'postcode' column at this stage (before: 0, after: 0), the code was kept as a preventive measure to ensure robustness in future runs.

Filling missing postcodes with -1 allows us to retain records without introducing nulls into models that require numerical inputs. Additionally, converting the column to integer format ensures consistent data types for analysis and machine learning pipelines.
"""
```

## C.5 Fixing Missing: "full_address"

```python
# <Student to fill this section>
missing_before = df_clean['full_address'].isnull().sum()

missing_full = df_clean['full_address'].isnull()
df_clean.loc[missing_full, 'full_address'] = (
    df_clean.loc[missing_full, 'street'] + " " +
    df_clean.loc[missing_full, 'type'] + ", " +
    df_clean.loc[missing_full, 'suburb'] + " " +
    df_clean.loc[missing_full, 'postcode'].astype(str)
)

missing_after = df_clean['full_address'].isnull().sum()

print(f"Missing values in 'full_address' before: {missing_before}, after: {missing_after}")
```

```python
df_clean['full_address'].isnull().sum()
```

```python
data_cleaning_extra_explanations = """
At this point in the workflow, there were no missing values in the 'full_address' column (before: 0, after: 0). This suggests that all addresses were already complete or had been handled earlier.

Nevertheless, the code to reconstruct full addresses was included as a precaution. It combines 'street', 'type', 'suburb', and 'postcode' into a structured format for any missing values that might occur in future data loads.

This makes the pipeline more robust and ensures address completeness without manual inspection.
"""
```

### D.4 New Feature "postcode_density_bin"

> Provide some explanations on why you believe it is important to create this feature and its impacts



```python
postcode_counts = df['postcode'].value_counts()
df['postcode_density'] = df['postcode'].map(postcode_counts)
df['postcode_density_bin'] = pd.qcut(df['postcode_density'], q=3, labels=['Low', 'Medium', 'High'])
postcode_density_output = df[['postcode', 'postcode_density', 'postcode_density_bin']].drop_duplicates().sort_values(by='postcode_density', ascending=False).head(10)

```

```python
print("\nTop Postcodes by Density Bin:")
print(postcode_density_output)
```

```python
# <Student to fill this section>
feature_engineering_3_explanations = """
We created a new categorical feature called 'postcode_density_bin' by segmenting postcodes into three bins — 'Low', 'Medium', and 'High' — based on the number of customers residing in each postcode.

For instance, postcode 2619 has 212 customers, while 2920 has 200, both falling under the 'High' density category. These high-density areas may represent key customer zones and could influence resource allocation, support coverage, or pricing models.

This binned feature simplifies numeric density values into interpretable segments that are useful for classification models or dashboards.
"""
```

```python
# Do not modify this code
print_tile(size="h3", key='feature_engineering_3_explanations', value=feature_engineering_3_explanations)
```
